```python
#funcoes definidas pelo usuario

''' Calculo do desconto '''
def calcular_desconto(valor, percentual):

    if percentual < 0 or percentual > 100:
        return None

    desconto = valor * (percentual / 100)
    return valor - desconto

''' Registrar vendas realiadas'''
def registrar_venda(produlto, valor_final):
 print(f">> venda regitrada: {produlto} por, R$ {valor_final:.2f}")

 #LAMBDA PARA ARRENDODAR
arredondar = lambda a: round(a,2)
```

```python
#lista de livros
class livro:   # criando o grupo livros
 def __init__(self, titulo, autor, genero, quantidade):
  self.titulo = titulo
  self.autor = autor
  self.genero = genero
  self.quantidade = quantidade
```

```python
livros = []
```

```python
#crinado uma definição para cadastras os livros no dicionario livros
def cadastrar_livro(titulo, autor, genero, quantidade):
 novo_livro = Livro(titulo, autor, genero, quantidade)
 livros.append(novo_livro)
 print(f"Livro '{titulo}' cadastrado com sucesso!")
```

```python
#cadastrar livrod usando a definição criada
cadastrar_livro("amor em ruinas ", "univer", "biblico", 30)
cadastrar_livro("nada a perder ", "Edir macedo", "biogrfia", 3)
cadastrar_livro("estrela ", "hollywod", "comedia", 15)
cadastrar_livro("reis", "univer", "biblico", 10)
```

```python
def catalogo(): # mostra os titulos dos livros dentro do dicionario livros
  print("\n--- CATALOGO DE LIVROS ---")
  for livro in livros:
    print(livro.titulo, livro.genero)
```

```python
livros_unicos = [] # deixando os livros sem duplicadas

for livro in livros:
    if livro.titulo not in [l.titulo for l in livros_unicos]:
        livros_unicos.append(livro)

livros = livros_unicos
```

```python
catalogo ()
```

```python
# cria uma busca dentro da lista declivro pelo titulo e retonar o achado
def buscar_livro(titulo):
    for livro in livros:
        if livro.titulo == titulo:
            print(f"Livro '{titulo}', encontrado")
            return True

    print(f"Livro '{titulo}', não encontrado")
    return False
```

```python
buscar_livro("amor em ruinas")
```

```python
#criando a lista de generos para relacionar com a quantidade

generos = {}
for livro in livros:
    if livro.genero in generos:
        generos[livro.genero] += livro.quantidade
    else:
        generos[livro.genero] = livro.quantidade
```

```python
# FASE 3 - unittest

import unittest


def calcular_total(valores):
    return sum(valores)


class TestCalcularTotal(unittest.TestCase):
    """Suíte de testes da função calcular_total."""

    def test_compra_normal(self):
        """Compra normal, soma de valores positivos."""
        self.assertEqual(calcular_total([10, 20, 30]), 60)

    def test_lista_vazia(self):
        """Lista sem itens deve retornar 0."""
        self.assertEqual(calcular_total([]), 0)

    def test_decimal(self):
        """Soma com valores em reais (FLOAT)."""
        self.assertAlmostEqual(
            calcular_total([29.90, 15.00, 5.50]),
            50.40,
            places=2
        )

    def test_um_item(self):
        """Lista com apenas um item retorna o valor do item."""
        self.assertEqual(calcular_total([10]), 10)

    def test_valores_negativos(self):
        """Reembolso: valores negativos cancelam positivos."""
        self.assertEqual(calcular_total([-10, 50]), 40)


# Rodando os testes no Jupyter/Colab
unittest.main(argv=[''], exit=False, verbosity=2)
```

```python
# Treinamento SUPRVISIONADO - regressao linear simples (y = 2X)
import tensorflow as tf
from tensorflow.keras.models import Sequential
from tensorflow.keras.layers import Dense
import numpy as np
import matplotlib.pyplot as plt
```

```python
#dados rotulados (entrada + saidas esperadas )
x_train = tf.constant([[1.0], [2.0], [3.0], [4.0]])
y_trayn = tf.constant([[2.0], [4.0], [6.0], [8.0]])
```

```python
model = Sequential()
model.add(Dense(units=1, input_shape=(1,)))
model.compile(optimizer='sgd', loss='mean_squared_error')
history = model.fit(x_train, y_trayn, epochs=1000, verbose=0)
```

```python
#previsao para um valor novo
x_new = tf.constant(  [[5.0]])
prediction = model.predict(x_new, verbose=0 )
print(f"Predição para x=5 : {prediction[0][0]:.2f} ( esperando : 10)")
```

```python

#2. Treinamento NÃO SUPERVISIONADO - autoencoder mínimo

import tensorflow as tf
from tensorflow.keras.layers import Input, Dense
from tensorflow.keras.models import Model

# Dados sem rótulo (só X)
X_unsupervised = tf.constant([[1.0, 2.0], [2.0, 3.0], [3.0, 4.0], [4.0, 5.0]])

# Autoencoder: encode comprime para 1 dimensão; decode reconstrói para 2
input_layer = Input(shape=(2,))
encoded = Dense(units=1)(input_layer)
decoded = Dense(units=2)(encoded)

autoencoder = Model(inputs=input_layer, outputs=decoded)
autoencoder.compile(optimizer='adam', loss='mean_squared_error')

# Treina tentando reconstruir a própria entrada
autoencoder.fit(X_unsupervised, X_unsupervised, epochs=1000, verbose=0)

prediction = autoencoder.predict(X_unsupervised, verbose=0)

print("Original:  ", X_unsupervised.numpy())
print("Reconstruído:", prediction)
```
